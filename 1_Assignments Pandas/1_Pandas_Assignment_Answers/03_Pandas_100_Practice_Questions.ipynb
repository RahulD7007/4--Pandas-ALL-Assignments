### Q56. Detect outliers in a column.
> 💡 **Hint:** Use the **IQR method**: compute Q1, Q3, IQR = Q3–Q1. Outliers are values below `Q1 - 1.5*IQR` or above `Q3 + 1.5*IQR`. Alternatively use Z-score: `|z| > 3`.

---

```python
# IQR Method:- (Skewed data)
Q1 = df["Marks"].quantile(0.25)
Q3 = df["Marks"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df["Marks"] < lower) |
    (df["Marks"] > upper)
]

outliers
```

```python
# z-score:- Normal Distribution
from scipy.stats import zscore

z = zscore(df["Marks"])

df[abs(z) > 3]
```

### Q98. Compare pivot vs groupby.

---

#### Pivot vs GroupBy
- Analysis -> groupby()
- reporting -> pivot_table()

| Feature | groupby() | pivot_table() |
| ------------------ | --------- | ------------- |
| Flexible | More Flexible | Moderate |
| 2D Report | No | Yes |
| Aggregation | Possible | Possible |
| Handles Duplicates | Yes | Yes |
| Spreadsheet Style  | No | Yes |
