# Series

A Series is a one-dimensional labelled array capable of holding heterogeneous data.

## Components
- Data
- Index
- Data Type

## Facts
- Mutable values
- Immutable size
- Supports vectorized operations

```python
# Create Series from list
s1 = pd.Series([10,20,30], index=['a','b','c'])
print(s1)
```

# DataFrame

A DataFrame is a two-dimensional labelled data structure.

## Properties
1. Heterogeneous Data
2. Size Mutable
3. Labelled Axes
4. Supports Missing Values
5. Powerful Indexing

```python
# List of lists
df1 = pd.DataFrame(
    [['John',25],['Mary',30]],
    columns=['Name','Age']
)
print(df1)
```

```python
# List of dictionaries
df2 = pd.DataFrame([
    {'Name':'Alice','Age':28},
    {'Name':'Bob','Age':35}
])
print(df2)
```

```python
# Dictionary of Series
df3 = pd.DataFrame({
    'A': pd.Series([1,2,3]),
    'B': pd.Series([4,5,6])
})
print(df3)
```

# Adding Columns

Methods:
- Assign scalar value
- Assign list
- Assign Series

```python
df = pd.DataFrame({
    'Name':['John','Mary','Alice'],
    'Age':[25,30,28]
})

# Scalar
df['Country'] = 'India'

# List
df['Score'] = [80,90,85]

# Series
bonus = pd.Series([5,10,15])
df['Bonus'] = bonus

print(df)
```

# Rename Columns

rename(columns={}, inplace=False)

Parameters:
- columns
- inplace

```python
# Rename single column
df.rename(columns={'Age':'Employee_Age'}, inplace=True)

# Rename multiple columns
df.rename(columns={
    'Score':'Exam_Score',
    'Bonus':'Extra_Bonus'
}, inplace=True)

print(df)
```

# Indexing and Data Selection

## Types
- loc (label based)
- iloc (position based)

```python
employees = pd.DataFrame({
    'Name':['John','Mary','Alice','Bob'],
    'Department':['IT','HR','IT','Finance'],
    'Salary':[50000,60000,55000,70000]
}, index=['E1','E2','E3','E4'])

print(employees)
```

## loc Examples

```python
# Access row
print(employees.loc['E1'])
print()

# Access column
print(employees.loc[:, 'Name'])

# Multiple rows and columns
print(employees.loc[['E1','E3'], ['Name','Salary']])

# Conditional indexing
print(employees.loc[employees['Salary'] > 55000])
```

## iloc Examples

```python
# Single row
print(employees.iloc[0])
print()

# Single column
print(employees.iloc[:,0])
print()

# Subset
print(employees.iloc[0:3, 0:2])
```

### 9. Difference between list of dictionaries and dictionary of Series?
---

- List of Dictionaries -> Row-oriented data (records).
- Dictionary of Series -> Column-oriented data (features)

#### Key Difference:-

| Feature | List of Dicts | Dict of Series |
|---------|-------------|--------------|
| Orientation | Rows | Columns | 
| Each Element Represents | One record | One column |
| Typical Source | APIs, JSON | Analytical datasets |
| Alignment | Row-based | Index-based |


---

```python
# List of dictionaries:- (Row - oriented)
data = [
    {'Name': 'Rahul', 'Age': 25},
    {'Name': 'Ganraj',  'Age': 30},
    {'Name': 'Neha',  'Age': 28}
]

df = pd.DataFrame(data)
df
```

```python
# Dictionary of Series:- Column-Oriented
data = {
    'Name': pd.Series(['Rahul', 'Ganraj', 'Neha']),
    'Age': pd.Series([25, 30, 28])
}

df = pd.DataFrame(data)
df
```

### 10. When would you choose loc over iloc?

- Use loc when selecting data using labels (row names or column names).
- Use iloc when selecting data using integer positions.

```python
df = pd.DataFrame({
    'Employee_ID': [101, 102, 103, 104],
    'Name': ['Rahul', 'Ganraj', 'Prasad', 'Neha'],
    'Salary': [50000, 70000, 45000, 80000]
})

df
```

```python
# using loc
df.loc[df['Salary'] > 60000]
```

```python
# iloc is wrong here 
# Because iloc only accepts integer positions, 
# not label-based boolean conditions.

# df.iloc[df['Salary'] > 60000]
```
