## Section 1: Basics & Series (Q1–Q20)

```python
# Q1. Create a Pandas Series from a list of integers
# pd.Series() converts a list into a labelled 1-D array; default index is 0, 1, 2, ...
s1 = pd.Series([10, 20, 30, 40, 50])
print(s1)
```

```python
# Q2. Create a Series from a dictionary with custom index
# Dict keys automatically become the index labels
s2 = pd.Series({'a': 100, 'b': 200, 'c': 300})
print(s2)
```

```python
# Q3. Access elements using index labels
# Single label, multiple labels, and label-based slice (inclusive on both ends)
s = pd.Series([10, 20, 30, 40], index=['a', 'b', 'c', 'd'])
print(s['a'])           # single element
print(s[['a', 'c']])    # multiple elements
print(s['a':'c'])       # slice — label-based, INCLUSIVE on both ends
```

```python
# Q4. Perform arithmetic operations on a Series
# Operations are element-wise and index-aligned automatically
s = pd.Series([1, 2, 3, 4, 5])
print('Add 10:', s + 10)
print('Multiply by 2:', s * 2)
print('Power of 2:', s ** 2)
print('Modulo 3:', s % 3)
```

```python
# Q5. Find mean, min, max of a Series
s = pd.Series([15, 42, 8, 93, 27, 56])
print('Mean   :', s.mean())
print('Median :', s.median())
print('Min    :', s.min())
print('Max    :', s.max())
print('Std Dev:', s.std())
print('Sum    :', s.sum())
```

```python
# Q6. Apply conditional filtering on a Series
# Boolean mask: True where condition holds, then select those positions
s = pd.Series([10, 55, 30, 80, 45, 90])
print('Values > 50:', s[s > 50])                     # single condition
print('Between 30 and 80:', s[(s >= 30) & (s <= 80)]) # combined with &
```

```python
# Q7. Replace values in a Series
# .replace() maps old value(s) to new value(s); pass a dict for multiple replacements
s = pd.Series([1, 2, 3, 2, 1, 4])
print(s.replace(1, 100))              # single replacement
print(s.replace({1: 100, 2: 200}))   # multiple replacements via dict
```

```python
# Q8. Convert Series data type
# .astype() casts to the specified dtype
s = pd.Series(['1', '2', '3', '4'])   # currently strings
print('Original dtype:', s.dtype)
s_int = s.astype(int)
print('After astype(int):', s_int.dtype)
s_float = s_int.astype(float)
print('After astype(float):', s_float.dtype)
```

```python
# Q9. Check for null values in Series
# .isnull() returns True where the value is NaN / None
s = pd.Series([1, None, 3, np.nan, 5])
print(s.isnull())           # boolean mask
print('Null count:', s.isnull().sum())
```

```python
# Q10. Fill missing values in Series
s = pd.Series([1, None, 3, np.nan, 5])
print('Fill with 0     :', s.fillna(0).tolist())
print('Fill with mean  :', s.fillna(s.mean()).tolist())
print('Forward fill    :', s.ffill().tolist())
print('Backward fill   :', s.bfill().tolist())
```

```python
# Q11. Create DataFrame from a list
# Each inner list becomes one row; provide column names separately
data = [[1, 'Alice', 85], [2, 'Bob', 90], [3, 'Charlie', 78]]
df11 = pd.DataFrame(data, columns=['ID', 'Name', 'Score'])
print(df11)
```

```python
# Q12. Create DataFrame from a dictionary
# Keys become column names; values (lists) must have equal length
df12 = pd.DataFrame({'Name': ['Alice', 'Bob', 'Charlie'],
                     'Age' : [25, 30, 35],
                     'City': ['Delhi', 'Mumbai', 'Pune']})
print(df12)
```

```python
# Q13. Create DataFrame from list of dictionaries
# Each dict is one row; missing keys automatically become NaN
records = [{'Name': 'Alice', 'Age': 25},
           {'Name': 'Bob',   'Age': 30, 'City': 'Mumbai'},
           {'Name': 'Charlie'}]
df13 = pd.DataFrame(records)
print(df13)
```

```python
# Q14. Display DataFrame structure (shape, columns, dtypes)
df = df12.copy()
print('Shape   :', df.shape)      # (rows, columns)
print('Columns :', df.columns.tolist())
print('Dtypes  :\n', df.dtypes)
print('\nInfo:')
df.info()
print('\nDescribe:\n', df.describe(include='all'))
```

```python
# Q15. Select single column from DataFrame → returns a Series
df = df12.copy()
print(df['Name'])           # bracket notation
# print(df.Name)           # dot notation also works (avoid for column names with spaces)
```

```python
# Q16. Select multiple columns → returns a DataFrame
# Use a list of column names inside double brackets
df = df12.copy()
print(df[['Name', 'City']])
```

```python
# Q17. Add new column with scalar value
# A single scalar is broadcast to every row automatically
df = df12.copy()
df['Country'] = 'India'
print(df)
```

```python
# Q18. Add column using Series
# Series index must align with the DataFrame index
df = df12.copy()
scores = pd.Series([88, 92, 75])   # index 0, 1, 2 aligns with df
df['Score'] = scores
print(df)
```

```python
# Q19. Rename columns using rename()
df = df12.copy()
df = df.rename(columns={'Name': 'Full_Name', 'Age': 'Years'})
print(df.columns.tolist())
```

## Section 2: Indexing, Selection & Slicing (Q21–Q40)

```python
# Shared sample DataFrame for Q21-Q40
df = pd.DataFrame({'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
                   'Dept': ['HR', 'IT', 'IT', 'Finance', 'HR'],
                   'Salary': [50000, 70000, 65000, 80000, 55000],
                   'Exp': [2, 5, 4, 7, 3]},
                  index=['e1', 'e2', 'e3', 'e4', 'e5'])
print(df)
```

```python
# Q21. Select rows using loc (label-based, INCLUSIVE on both ends)
print(df.loc['e1'])              # single row → Series
print(df.loc['e2':'e4'])         # range of rows
```

```python
# Q22. Select rows using iloc (position-based, EXCLUSIVE on the end)
print(df.iloc[0])          # first row
print(df.iloc[1:4])        # rows at positions 1, 2, 3 (not 4)
```

```python
# Q23. Access a specific row and column (single cell)
print(df.loc['e1', 'Name'])     # label-based
print(df.iloc[0, 0])            # position-based
```

```python
# Q24. Select subset using multiple rows & columns
print(df.loc[['e1', 'e3'], ['Name', 'Salary']])   # label-based
print(df.iloc[[0, 2], [0, 2]])                     # position-based
```

```python
# Q25. Conditional filtering using loc
# Pass a boolean Series as the row selector; select specific columns
print(df.loc[df['Salary'] > 60000, ['Name', 'Salary']])
```

```python
# Q26. Slice rows using index range
print(df.iloc[1:4])           # positions 1, 2, 3
print(df.loc['e2':'e4'])      # labels e2, e3, e4 (all inclusive)
```

```python
# Q27. Slice columns using labels
# loc[:, 'col1':'col3'] — all rows, column range (label-inclusive)
print(df.loc[:, 'Name':'Salary'])
```

```python
# Q28. Slice columns using position
# iloc[:, 1:4] — all rows, column positions 1, 2, 3
print(df.iloc[:, 1:4])
```

```python
# Q29. Combined row & column slicing
print(df.loc[df['Exp'] >= 4, ['Name', 'Dept', 'Salary']])  # label-based
print(df.iloc[0:3, [0, 2]])                                 # position-based
```

```python
# Q30. Set a column as index
df_idx = df.reset_index(drop=True)   # start from a fresh integer index
df_idx = df_idx.set_index('Name')    # 'Name' becomes the row index
print(df_idx)
```

```python
# Q31. Create hierarchical index (MultiIndex)
# Passing a list of columns creates a two-level (or more) index
df_mi = df.reset_index(drop=True).set_index(['Dept', 'Name'])
print(df_mi)
print(type(df_mi.index))
```

```python
# Q32. Reset index — moves current index back to a regular column
df_reset = df.reset_index()   # 'e1','e2'... move to a column named 'index'
print(df_reset)
```

```python
# Q33. Reset multi-index — flattens both index levels back to columns
df_reset_mi = df_mi.reset_index()
print(df_reset_mi.head())
```

```python
# Q34. Convert index to column (default behaviour of reset_index)
# drop=False (default) keeps the index as a new column
df_temp = df.copy()
df_temp.index.name = 'EmpID'
print(df_temp.reset_index())   # EmpID becomes a regular column
```

```python
# Q35. Use drop=True in reset_index to discard the old index
print(df.reset_index(drop=True))   # original 'e1','e2'... labels are thrown away
```

```python
# Q36. Label-based indexing with loc
print(df.loc['e3'])               # works for any index type (string, int, datetime)
```

```python
# Q37. Position-based indexing with iloc
# Always uses integer positions regardless of what the index labels are
print(df.iloc[2])   # third row (0-indexed)
```

```python
# Q38. Access first N rows
print(df.head(3))
```

```python
# Q39. Access last N rows
print(df.tail(3))
```

```python
# Q40. Extract random sample from DataFrame
print(df.sample(n=3, random_state=42))          # fixed seed for reproducibility
print(df.sample(frac=0.4, random_state=42))     # 40 % of rows
```

## Section 3: Data Cleaning (Q41–Q60)

```python
# Shared messy DataFrame for Q41-Q60
df_raw = pd.DataFrame({
    'Name'  : ['  Alice', 'Bob', None, 'Diana', 'bob', 'Eve'],
    'Age'   : [25, 30, None, 35, 30, 'N/A'],
    'Salary': [50000, None, 65000, 80000, None, 55000],
    'City'  : ['Delhi', 'Mumbai', 'Pune', None, 'Mumbai', 'Delhi']
})
print(df_raw)
```

```python
# Q41. Detect missing values using isnull()
print(df_raw.isnull())
print('\nColumn-wise null count:\n', df_raw.isnull().sum())
```

```python
# Q42. Drop missing values using dropna()
print('Drop any row with NaN:\n', df_raw.dropna())
print('\nDrop rows where Salary is NaN:\n', df_raw.dropna(subset=['Salary']))
print('\nKeep rows with at least 3 non-null values:\n', df_raw.dropna(thresh=3))
```

```python
# Q43. Fill missing values with a constant
df_filled = df_raw.fillna({'Name': 'Unknown', 'City': 'Unknown', 'Salary': 0})
print(df_filled)
```

```python
# Q44. Fill missing values using mean/median
df_num = df_raw.copy()
df_num['Salary'] = pd.to_numeric(df_num['Salary'], errors='coerce')
df_num['Salary'] = df_num['Salary'].fillna(df_num['Salary'].mean())
print(df_num['Salary'])
```

```python
# Q45. Use forward fill and backward fill
s_gaps = pd.Series([1, None, None, 4, None, 6])
print('Forward fill :', s_gaps.ffill().tolist())
print('Backward fill:', s_gaps.bfill().tolist())
```

```python
# Q46. Convert data type using astype()
df_c = pd.DataFrame({'Score': ['85', '90', '78', '92']})
print('Before:', df_c['Score'].dtype)
df_c['Score'] = df_c['Score'].astype(int)
print('After :', df_c['Score'].dtype)
```

```python
# Q47. Clean string column using .str.lower()
# The .str accessor exposes vectorised string methods
df_str = pd.DataFrame({'City': ['Delhi', 'MUMBAI', 'Pune', 'BANGALORE']})
df_str['City'] = df_str['City'].str.lower()
print(df_str)
```

```python
# Q48. Trim whitespace in strings
df_ws = pd.DataFrame({'Name': ['  Alice', 'Bob  ', '  Charlie  ']})
df_ws['Name'] = df_ws['Name'].str.strip()   # removes both leading & trailing spaces
print(df_ws)
```

```python
# Q49. Replace substring in column
df_ph = pd.DataFrame({'Phone': ['123-456-7890', '987-654-3210']})
df_ph['Phone'] = df_ph['Phone'].str.replace('-', '', regex=False)
print(df_ph)
```

```python
# Q50. Extract part of string using .str
df_em = pd.DataFrame({'Email': ['alice@gmail.com', 'bob@yahoo.com', 'charlie@outlook.com']})
df_em['Username'] = df_em['Email'].str.split('@').str[0]    # before @
df_em['Domain']   = df_em['Email'].str.split('@').str[1]    # after @
print(df_em)
```

```python
# Q51. Detect duplicate rows
df_dup = pd.DataFrame({'A': [1, 2, 2, 3], 'B': ['x', 'y', 'y', 'z']})
print(df_dup.duplicated())            # boolean mask
print(df_dup[df_dup.duplicated()])    # view duplicate rows
```

```python
# Q52. Remove duplicates using drop_duplicates()
df_dup2 = pd.DataFrame({'Name': ['Alice', 'Bob', 'Alice', 'Charlie', 'Bob'],
                         'Score': [85, 90, 85, 78, 90]})
print('Keep first occurrence:\n', df_dup2.drop_duplicates())
print('Keep last occurrence:\n',  df_dup2.drop_duplicates(keep='last'))
print('Drop ALL duplicates:\n',   df_dup2.drop_duplicates(keep=False))
```

```python
# Q53. Rename row labels
df_idx2 = pd.DataFrame({'Val': [10, 20, 30]}, index=['a', 'b', 'c'])
df_idx2 = df_idx2.rename(index={'a': 'row1', 'b': 'row2', 'c': 'row3'})
print(df_idx2)
```

```python
# Q54. Handle inconsistent data
# Normalise strings: strip whitespace, lowercase, then map to a canonical form
df_inc = pd.DataFrame({'Gender': ['Male', ' male', 'M', 'Female', 'female', 'F']})
df_inc['Gender'] = (df_inc['Gender']
                    .str.strip()
                    .str.lower()
                    .replace({'m': 'male', 'f': 'female'}))
print(df_inc)
```

```python
# Q55. Replace values in DataFrame
df_rep = pd.DataFrame({'Status': ['Active', 'Inactive', 'Pending', 'Active']})
print(df_rep.replace('Active', 'A'))                          # global replace
print(df_rep.replace({'Status': {'Active': 'A', 'Inactive': 'I'}}))  # column-specific
```

```python
# Q56. Detect outliers using the IQR method
df_out = pd.DataFrame({'Score': [10, 20, 15, 200, 18, 22, 19, -100, 21]})
Q1  = df_out['Score'].quantile(0.25)
Q3  = df_out['Score'].quantile(0.75)
IQR = Q3 - Q1
lower, upper = Q1 - 1.5 * IQR, Q3 + 1.5 * IQR
outliers = df_out[(df_out['Score'] < lower) | (df_out['Score'] > upper)]
print(f'Q1={Q1}, Q3={Q3}, IQR={IQR}')
print('Outliers:\n', outliers)
```

```python
# Q57. Handle outliers using clipping
# .clip() replaces values outside [lower, upper] with the boundary values
df_out['Score_Clipped'] = df_out['Score'].clip(lower=lower, upper=upper)
print(df_out)
```

```python
# Q58. Standardise column names
# Chain: strip whitespace → lowercase → replace spaces with underscores
df_cols = pd.DataFrame(columns=['First Name', ' Age ', 'Salary (INR)', 'City '])
df_cols.columns = df_cols.columns.str.strip().str.lower().str.replace(' ', '_').str.replace('(', '').str.replace(')', '')
print(df_cols.columns.tolist())
```

```python
# Q59. Handle mixed data types in a column
# pd.to_numeric with errors='coerce' converts non-numeric to NaN
df_mix = pd.DataFrame({'Value': ['10', '20', 'abc', '40', None, '50']})
df_mix['Value'] = pd.to_numeric(df_mix['Value'], errors='coerce')
print(df_mix)
print(df_mix['Value'].dtype)
```

```python
# Q60. Remove special characters from text data using regex
df_spec = pd.DataFrame({'Comment': ['Hello!', 'Good @work', 'Nice #1', 'No$special%']})
df_spec['Comment_Clean'] = df_spec['Comment'].str.replace(r'[^a-zA-Z0-9 ]', '', regex=True)
print(df_spec)
```

## Section 4: Data Manipulation (Q61–Q80)

```python
# Shared sample DataFrame for Q61-Q80
df_emp = pd.DataFrame({
    'Name'  : ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve', 'Frank'],
    'Dept'  : ['HR', 'IT', 'IT', 'Finance', 'HR', 'Finance'],
    'Salary': [50000, 70000, 65000, 80000, 55000, 72000],
    'Exp'   : [2, 5, 4, 7, 3, 6]
})
print(df_emp)
```

```python
# Q61. Filter rows using AND condition (&)
# Each condition must be wrapped in parentheses
result = df_emp[(df_emp['Salary'] > 60000) & (df_emp['Dept'] == 'IT')]
print(result)
```

```python
# Q62. Filter rows using OR condition (|)
result = df_emp[(df_emp['Dept'] == 'HR') | (df_emp['Salary'] > 70000)]
print(result)
```

```python
# Q63. Filter rows using NOT condition (~)
# ~ negates the boolean mask
result = df_emp[~(df_emp['Dept'] == 'IT')]   # everyone NOT in IT
print(result)
```

```python
# Q64. Sort DataFrame by column values
print(df_emp.sort_values('Salary', ascending=False))              # single column descending
print(df_emp.sort_values(by=['Dept', 'Salary'], ascending=[True, False]))  # multi-column
```

```python
# Q65. Sort DataFrame by index
df_rev = df_emp.iloc[::-1]         # reverse order first
print(df_rev.sort_index())         # sort by integer index ascending
# sort columns alphabetically: df.sort_index(axis=1)
```

```python
# Q66. Add a calculated column
df_emp['Annual_Bonus'] = df_emp['Salary'] * 0.10   # direct arithmetic
# Chainable alternative: df_emp.assign(Annual_Bonus=lambda x: x['Salary'] * 0.10)
print(df_emp)
```

```python
# Q67. Remove column(s) using drop()
df_d = df_emp.copy()
df_d = df_d.drop(columns=['Annual_Bonus'])
print(df_d.columns.tolist())
```

```python
# Q68. Remove rows using drop()
df_dr = df_emp.copy()
df_dr = df_dr.drop(index=[0, 2])   # drop rows with index labels 0 and 2
print(df_dr)
```

```python
# Q69. Use map() on a column
# map() works element-wise; ideal for dictionary-based substitution
df_m = df_emp.copy()
dept_code = {'HR': 'H', 'IT': 'I', 'Finance': 'F'}
df_m['Dept_Code'] = df_m['Dept'].map(dept_code)
print(df_m[['Name', 'Dept', 'Dept_Code']])
```

```python
# Q70. Apply function using apply()
# apply() on a Series: element-wise; on DataFrame: row- or column-wise
df_a = df_emp.copy()
df_a['Grade'] = df_a['Salary'].apply(lambda x: 'High' if x >= 70000 else 'Mid' if x >= 55000 else 'Low')
print(df_a[['Name', 'Salary', 'Grade']])
```

```python
# Q71. Apply element-wise using DataFrame.map() (pandas >= 2.1)
# For older pandas use .applymap(); both apply a function to every element
df_num_only = df_emp[['Salary', 'Exp']].copy()
# Multiply every element by 2 just to illustrate
print(df_num_only.map(lambda x: x * 2))
```

```python
# Q72. Chain multiple operations
# Wrap in parentheses for multi-line readability
result = (
    df_emp.copy()
    .loc[df_emp['Salary'] >= 60000]
    .sort_values('Salary', ascending=False)
    [['Name', 'Dept', 'Salary']]
    .assign(Salary_K=lambda x: (x['Salary'] / 1000).round(1))
    .reset_index(drop=True)
)
print(result)
```

```python
# Q73. Normalise numeric columns (Min-Max scaling → range [0, 1])
col = df_emp['Salary']
df_emp['Salary_Norm'] = (col - col.min()) / (col.max() - col.min())
print(df_emp[['Name', 'Salary', 'Salary_Norm']])
```

```python
# Q74. Scale numerical data (Z-score standardisation → mean=0, std=1)
col = df_emp['Salary']
df_emp['Salary_Z'] = (col - col.mean()) / col.std()
print(df_emp[['Name', 'Salary', 'Salary_Z']].round(3))
```

```python
# Q75. Rank values in a column
df_emp['Salary_Rank'] = df_emp['Salary'].rank(ascending=False, method='dense').astype(int)
print(df_emp[['Name', 'Salary', 'Salary_Rank']].sort_values('Salary_Rank'))
```

```python
# Q76. Create bins using cut() — equal-width intervals
df_emp['Exp_Bin'] = pd.cut(df_emp['Exp'], bins=[0, 3, 6, 9],
                            labels=['Junior', 'Mid', 'Senior'])
print(df_emp[['Name', 'Exp', 'Exp_Bin']])
```

```python
# Q77. Create bins using qcut() — equal-frequency (quantile-based) bins
df_emp['Salary_Quartile'] = pd.qcut(df_emp['Salary'], q=4,
                                     labels=['Q1', 'Q2', 'Q3', 'Q4'],
                                     duplicates='drop')
print(df_emp[['Name', 'Salary', 'Salary_Quartile']])
```

```python
# Q78. Count unique values
print('Unique depts  :', df_emp['Dept'].nunique())
print('\nFrequency per dept:\n', df_emp['Dept'].value_counts())
```

```python
# Q79. Replace multiple values at once
df_rep2 = df_emp[['Dept']].copy()
print(df_rep2.replace({'Dept': {'HR': 'Human Resources', 'IT': 'Information Technology'}}))
```

```python
# Q80. Create a flag column based on a condition
# Cast boolean to int: True→1, False→0
df_emp['High_Earner'] = (df_emp['Salary'] >= 70000).astype(int)
print(df_emp[['Name', 'Salary', 'High_Earner']])
```

## Section 5: Iteration, GroupBy & Pivot (Q81–Q100)

```python
# Shared clean DataFrame for Q81-Q100
df_sales = pd.DataFrame({
    'Region'  : ['North', 'South', 'North', 'East', 'South', 'East', 'North', 'South'],
    'Product' : ['A', 'B', 'B', 'A', 'A', 'B', 'A', 'B'],
    'Sales'   : [200, 150, 300, 250, 180, 220, 210, 170],
    'Units'   : [20, 15, 30, 25, 18, 22, 21, 17]
})
print(df_sales)
```

```python
# Q81. Iterate over rows using iterrows()
# Returns (index, Series) pairs — note: can be slow on large DataFrames
for idx, row in df_sales.head(3).iterrows():
    print(f'Index={idx}, Region={row["Region"]}, Sales={row["Sales"]}')
```

```python
# Q82. Iterate using itertuples() — faster than iterrows()
# Returns named tuples; access fields via row.ColumnName
for row in df_sales.head(3).itertuples():
    print(f'Index={row.Index}, Region={row.Region}, Sales={row.Sales}')
```

```python
# Q83. Iterate over columns using df.items()
# Yields (column_name, Series) pairs
for col_name, col_data in df_sales.items():
    print(f'Column: {col_name} | dtype: {col_data.dtype}')
```

```python
# Q84. Avoid iteration using vectorisation
# SLOW (loop): for i, row in df.iterrows(): df.at[i, 'Revenue'] = row['Sales'] * 1.2
# FAST (vectorised): direct column arithmetic — much faster
df_sales['Revenue'] = df_sales['Sales'] * 1.2
print(df_sales.head(3))
```

```python
# Q85. Group data using groupby() and compute mean
print(df_sales.groupby('Region')['Sales'].mean())
```

```python
# Q86. Calculate sum per group
print(df_sales.groupby('Region')['Sales'].sum())
```

```python
# Q87. Calculate multiple aggregations with agg()
result = df_sales.groupby('Region').agg(
    Total_Sales=('Sales', 'sum'),
    Avg_Sales  =('Sales', 'mean'),
    Max_Units  =('Units', 'max')
)
print(result)
```

```python
# Q88. Group by multiple columns
print(df_sales.groupby(['Region', 'Product'])['Sales'].sum())
```

```python
# Q89. Filter grouped data — keep only groups where mean Sales > 200
filtered = df_sales.groupby('Region').filter(lambda g: g['Sales'].mean() > 200)
print(filtered)
```

```python
# Q90. Transform grouped data using transform()
# Returns a Series same length as original df, filled with group statistics
df_sales['Region_Avg_Sales'] = df_sales.groupby('Region')['Sales'].transform('mean')
print(df_sales[['Region', 'Sales', 'Region_Avg_Sales']])
```

```python
# Q91. Apply a custom function in groupby
# Lambda receives a Series for each group; must return a scalar
result = df_sales.groupby('Product')['Sales'].agg(lambda s: s.max() - s.min())
print('Sales range per product:\n', result)
```

```python
# Q92. Create a basic pivot table
pivot = pd.pivot_table(df_sales, values='Sales', index='Region',
                       columns='Product', aggfunc='mean')
print(pivot)
```

```python
# Q93. Use multiple aggregation functions in pivot table
pivot_multi = pd.pivot_table(df_sales, values='Sales', index='Region',
                              columns='Product', aggfunc=['mean', 'max'])
print(pivot_multi)
```

```python
# Q94. Handle missing values in pivot tables with fill_value
# Some Region-Product combos may not exist → NaN; fill_value replaces them
pivot_fill = pd.pivot_table(df_sales, values='Sales', index='Region',
                             columns='Product', aggfunc='sum', fill_value=0)
print(pivot_fill)
```

```python
# Q95. Add margins (row/column totals) to pivot table
pivot_margins = pd.pivot_table(df_sales, values='Sales', index='Region',
                                columns='Product', aggfunc='sum',
                                fill_value=0,
                                margins=True, margins_name='TOTAL')
print(pivot_margins)
```

```python
# Q97. Convert DataFrame to pivot format using df.pivot()
# Requires unique (index, columns) key pairs — no aggregation performed
df_unique = pd.DataFrame({
    'Region' : ['North', 'North', 'South', 'South'],
    'Product': ['A', 'B', 'A', 'B'],
    'Sales'  : [200, 300, 180, 150]
})
print(df_unique.pivot(index='Region', columns='Product', values='Sales'))
```

```python
# Q98. Compare pivot vs groupby
# groupby: flat structure, flexible, handles duplicates
gb_result = df_sales.groupby(['Region', 'Product'])['Sales'].sum().unstack()
print('GroupBy result:\n', gb_result)

# pivot_table: 2D grid, handles duplicates via aggfunc
pt_result = pd.pivot_table(df_sales, values='Sales',
                            index='Region', columns='Product', aggfunc='sum', fill_value=0)
print('\nPivot Table result:\n', pt_result)

# pivot: no aggregation → fails with duplicates (works with df_unique)
pv_result = df_unique.pivot(index='Region', columns='Product', values='Sales')
print('\nPivot result (unique keys):\n', pv_result)
```

```python
# Q100. End-to-end analysis workflow
# Step 1: Create / load data
data_e2e = pd.DataFrame({
    'Name'   : ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve', 'Frank', 'Alice', None],
    'Dept'   : ['HR', 'IT', 'IT', 'Finance', 'HR', 'Finance', 'HR', 'IT'],
    'Salary' : [50000, 70000, 65000, 80000, 55000, 72000, 50000, None],
    'Exp'    : [2, 5, 4, 7, 3, 6, 2, None]
})

# Step 2: Inspect
print('Shape:', data_e2e.shape)
print('Nulls:\n', data_e2e.isnull().sum())

# Step 3: Clean
data_e2e = data_e2e.dropna(subset=['Name'])              # remove row with no Name
data_e2e['Salary'] = data_e2e['Salary'].fillna(data_e2e['Salary'].median())
data_e2e['Exp']    = data_e2e['Exp'].fillna(data_e2e['Exp'].median())
data_e2e = data_e2e.drop_duplicates()

# Step 4: Feature engineering
data_e2e['Salary_K']   = (data_e2e['Salary'] / 1000).round(1)
data_e2e['Exp_Level']  = pd.cut(data_e2e['Exp'], bins=[0,3,6,10],
                                  labels=['Junior','Mid','Senior'])

# Step 5: GroupBy / pivot analysis
dept_summary = data_e2e.groupby('Dept').agg(
    Headcount=('Name', 'count'),
    Avg_Salary=('Salary', 'mean'),
    Total_Salary=('Salary', 'sum')
).round(2)
print('\nDept Summary:\n', dept_summary)

# Step 6: Sort and rank
dept_summary = dept_summary.sort_values('Avg_Salary', ascending=False)
dept_summary['Rank'] = dept_summary['Avg_Salary'].rank(ascending=False).astype(int)
print('\nRanked:\n', dept_summary)

# Step 7: Export (commented out to avoid file creation)
# dept_summary.to_csv('dept_report.csv')
print('\n End-to-end workflow complete!')
```
