## 2. Introduction to Pandas

### What is Pandas?
**Pandas** (Panel Data) is an open-source Python library built on top of NumPy.  
It provides **fast, flexible, and expressive data structures** designed to make working with structured (tabular, multidimensional, potentially heterogeneous) and time series data both easy and intuitive.

### Core Functionality
| Feature | Description |
|---------|-------------|
| Data Loading | Read/write CSV, Excel, JSON, SQL, Parquet, etc. |
| Data Cleaning | Handle missing values, duplicates, type casting |
| Data Manipulation | Filter, sort, group, merge, reshape |
| Statistical Analysis | describe(), mean(), std(), corr() |
| Time Series | Datetime indexing, resampling, rolling windows |

### Key Data Structures
| Structure | Dimensions | Description |
|-----------|-----------|-------------|
| **Series** | 1D | Labeled array holding any data type |
| **DataFrame** | 2D | Table of rows × columns (most used) |

````python
# Example: Plain Python vs Pandas
# ---- Plain Python ----
students = [("Alice", 85), ("Bob", 92), ("Charlie", 78)]
avg = sum(s[1] for s in students) / len(students)
print("Plain Python average:", avg)

# ---- Pandas way ----
df_students = pd.DataFrame(students, columns=["Name", "Score"])
print("\nPandas DataFrame:")
print(df_students)
print(f"\nMean Score (pandas): {df_students['Score'].mean()}")
print(f"Max  Score (pandas): {df_students['Score'].max()}")
````

## 4. Series — 1D Labeled Array

### What is a Series?
A **Series** is a **one-dimensional labeled array** capable of holding any data type  
(integers, strings, floats, Python objects, etc.).

```
Index  →  Labels (default: 0, 1, 2 …)
Values →  The actual data
```

### Components of a Series
| Component | Description |
|-----------|-------------|
| `data`    | The array-like values |
| `index`   | Labels for each element (default RangeIndex) |
| `dtype`   | The data type of the values |
| `name`    | Optional name for the Series |

### Key Facts
- Default index is `0, 1, 2, …` (integer-based)
- Can hold **any Python data type**
- Supports **label-based access**: `s['label']`
- A DataFrame column is a Series

## 5. Creating a Series

````python
# ── Series components ─────────────────────────────────────────────

s = pd.Series([10, 20, 30, 40, 50], name="Marks")

print("Series:\n", s)
print("\nIndex :", s.index.tolist())
print("\nValues:", s.values)
print("\ndtype :", s.dtype)
print("\nname  :", s.name)
````

````python
# ── Create from list with default index ───────────────────────────
print("\n--- From list (default index) ---")

s1 = pd.Series([100, 200, 300, 400])
print(s1)
````

````python
# ── Create from list with custom index ───────────────────────────
print("\n--- From list (custom index) ---")

s2 = pd.Series([100, 200, 300, 400], index=["a", "b", "c", "d"])
print(s2)
print("Access by label s2['b']:", s2["b"])
````

````python
# ── Create from dictionary ────────────────────────────────────────
print("\n--- From dictionary (keys become index) ---")

marks_dict = {"Math": 90, "Science": 85, "English": 78, "History": 92}

s3 = pd.Series(marks_dict)
print(s3)
print("Access Science:", s3["Science"])
````

````python
# ── Create from scalar ────────────────────────────────────────────
print("\n--- From scalar (broadcasts to all index labels) ---")

s4 = pd.Series(5, index=["x", "y", "z"])
print(s4)
````

````python
# ── Index alignment ───────────────────────────────────────────────
print("\n--- Index alignment example ---")

s_a = pd.Series({"Math": 90, "Science": 85, "English": 78})
s_b = pd.Series({"Math": 80, "Science": 95, "History": 70})
print("Sum (NaN where index doesn't match):\n", s_a + s_b)
````

## 6. DataFrame — Introduction and Structure

### What is a DataFrame?
A **DataFrame** is a **2-dimensional labeled data structure** with columns of potentially different data types.  
Think of it as a spreadsheet, SQL table, or a dictionary of Series objects.

```
         Name   Age  Score
    0   Alice    25   88.5      ← Row (axis=0)
    1     Bob    30   92.0
    2  Charlie   22   75.0
    ↑
 Column (axis=1)
```

### Attributes of a DataFrame
| Attribute | Description |
|-----------|-------------|
| `df.shape`   | (rows, columns) |
| `df.dtypes`  | Data type of each column |
| `df.columns` | Column labels |
| `df.index`   | Row labels |
| `df.values`  | Pandas array of values |
| `df.info()`  | Summary of dtypes & non-null counts |
| `df.describe()` | Statistical summary |

````python
df = pd.DataFrame({
    "Name" : ["Alice", "Bob", "Charlie", "Diana"],
    "Age"  : [25, 30, 22, 28],
    "Score": [88.5, 92.0, 75.0, 95.5],
    "City" : ["NY", "LA", "NY", "Chicago"]
})

print("DataFrame:\n", df)
print("\nshape   :", df.shape)
print("\ncolumns :", df.columns.tolist())
print("\nindex   :", df.index.tolist())
print("\ndtypes:\n", df.dtypes)
print("\ninfo():")
df.info()
print("\ndescribe():\n", df.describe())
````

## 7. Properties of DataFrame

### 1. Heterogeneous Data — different dtypes in different columns
### 2. Size-Mutable — rows and columns can be added/removed
### 3. Labeled Axes — named rows and columns
### 4. Alignment — operations auto-align on index
### 5. Integration — interoperates with NumPy

````python
# ── 1. Heterogeneous Data ─────────────────────────────────────────
df_hetero = pd.DataFrame({
    "Name"   : ["Alice", "Bob"],     # object (string)
    "Age"    : [25, 30],             # int64
    "Score"  : [88.5, 92.0],         # float64
    "Passed" : [True, True]          # bool
})
print("1. Heterogeneous Data:")
print(df_hetero.dtypes, "\n")
````

````python
# ── 2. Size-Mutable ───────────────────────────────────────────────
df_mut = df_hetero.copy()

df_mut["Grade"] = ["A", "A+"]         # add column
print("2. Size-Mutable — after adding 'Grade' column:")
print(df_mut)

df_mut = df_mut.drop(columns=["Grade"])  # Drop
print("   After removing 'Grade':", df_mut.columns.tolist(), "\n")
````

````python
# ── 3. Labeled Axes ───────────────────────────────────────────────
df_labeled = pd.DataFrame(
    {"Math": [90, 85], "Science": [78, 92]},
    index=["Alice", "Bob"]
)
print("3. Labeled Axes (custom row index):")
print(df_labeled, "\n")
````

````python
# ── 4. Alignment ──────────────────────────────────────────────────
s1 = pd.Series({"Math": 10, "Science": 20, "English": 30})
s2 = pd.Series({"Math":  5, "Science": 15, "History": 25})
print("4. Alignment — adding two Series with different indices:")
print(s1 + s2, "\n")   # NaN where indices don't overlap
````

````python
# ── 5. Integration with NumPy ─────────────────────────────────────
df_int = pd.DataFrame({"A": [1, 2, 3], "B": [4, 5, 6]})
print("5. Integration — applying np.sqrt on DataFrame values:")
print(np.sqrt(df_int))
````

## 8. Creating a DataFrame

Three common ways:
1. **List of Lists** — specify column names explicitly
2. **List of Dictionaries** — keys become column names
3. **Dictionary of Series** — index alignment is automatic

````python
# ── Method 1: List of Lists ───────────────────────────────────────
data_lol = [
    ["Alice",   25, 88.5],
    ["Bob",     30, 92.0],
    ["Charlie", 22, 75.0],
    ["Diana",   28, 95.5],
]

df1 = pd.DataFrame(data_lol, columns=["Name", "Age", "Score"])
print("Method 1 — List of Lists:")
print(df1)
````

````python
# ── Method 2: List of Dictionaries ───────────────────────────────
data_lod = [
    {"Name": "Alice",   "Age": 25, "Score": 88.5},
    {"Name": "Bob",     "Age": 30, "Score": 92.0},
    {"Name": "Charlie", "Age": 22, "Score": 75.0},
    {"Name": "Diana",   "Age": 28},                # missing Score → NaN
]

df2 = pd.DataFrame(data_lod)
print("\nMethod 2 — List of Dictionaries (note NaN for Diana's Score):")
print(df2)
````

````python
# ── Method 3: Dictionary of Series ───────────────────────────────
names  = pd.Series(["Alice", "Bob", "Charlie", "Diana"])
ages   = pd.Series([25, 30, 22, 28])
scores = pd.Series([88.5, 92.0, 75.0, 95.5])

df3 = pd.DataFrame({"Name": names, "Age": ages, "Score": scores})
print("\nMethod 3 — Dictionary of Series:")
print(df3)
````

````python
# Index alignment demo with Series
s_a = pd.Series([1, 2, 3], index=[0, 1, 2])
s_b = pd.Series([10, 20],  index=[0, 1])      # shorter series
df_align = pd.DataFrame({"A": s_a, "B": s_b})
print("\nIndex alignment — shorter Series gets NaN:")
print(df_align)
````

## 9. Adding Columns to a DataFrame

| Method | Description |
|--------|-------------|
| `df['new_col'] = scalar` | Broadcasts one value to all rows |
| `df['new_col'] = list/Series` | Assigns element-wise |
| `df.assign(col=...)` | Returns a new DataFrame (chainable) |

````python
df_add = pd.DataFrame({
    "Name" : ["Alice", "Bob", "Charlie", "Diana"],
    "Score": [88.5, 92.0, 75.0, 95.5]
})

# ── Assign scalar value (broadcasts to all rows) ──────────────────
df_add["Country"] = "India"
print("After scalar assignment (Country = India):")
print(df_add)
````

````python
# ── Assign a list ─────────────────────────────────────────────────

df_add["Grade"] = ["A", "A+", "B+", "A+"]

print("\nAfter list assignment (Grade):")
print(df_add)
````

````python
# ── Assign a Series (index-aligned) ──────────────────────────────

bonus = pd.Series([5.0, 3.0, 7.0, 2.0], index=[0, 1, 2, 3])
df_add["Bonus"] = bonus
print("\nAfter Series assignment (Bonus):")
print(df_add)
````

````python
# ── Using assign() — returns new DataFrame ────────────────────────

df_new = df_add.assign(
    FinalScore = df_add["Score"] + df_add["Bonus"],
    Passed     = lambda x: x["FinalScore"] >= 80
)
print("\nUsing assign() — chained column creation:")
print(df_new)
````

## 10. Renaming Columns — `rename()`

**Syntax:** `df.rename(columns={old: new}, inplace=False)`

| Parameter | Description |
|-----------|-------------|
| `columns` | Dict mapping old names → new names |
| `index`   | Dict mapping old row labels → new labels |
| `inplace` | `True` = modify in place; `False` (default) = return new DataFrame |

````python
df_ren = pd.DataFrame({
    "nm" : ["Alice", "Bob", "Charlie"],
    "ag" : [25, 30, 22],
    "sc" : [88.5, 92.0, 75.0],
    "ct" : ["NY", "LA", "Chicago"]
})
print("Original columns:", df_ren.columns.tolist())

# ── Rename single column (returns new df) ─────────────────────────
df_r1 = df_ren.rename(columns={"nm": "Name"})
print("\nRename single column 'nm' → 'Name':")
print(df_r1.columns.tolist())
````

````python
# ── Rename multiple columns ───────────────────────────────────────
df_r2 = df_ren.rename(columns={"nm": "Name", "ag": "Age",
                                "sc": "Score", "ct": "City"})
print("\nRename multiple columns:")
print(df_r2)
````

````python
# ── inplace=True ──────────────────────────────────────────────────
df_r3 = df_ren.copy()
df_r3.rename(columns={"nm": "Name", "ag": "Age"}, inplace=True)
print("\nAfter inplace=True rename:")
print(df_r3.columns.tolist())
````

## 11. Indexing and Data Selection

### Types of Indexing
| Indexer | Basis | Usage |
|---------|-------|-------|
| `loc`   | **Label-based** | Use actual row/column names |
| `iloc`  | **Position-based** | Use integer positions (0-based) |

> **Rule of Thumb:** Use `loc` when you know the label. Use `iloc` when you know the position.

---
### `loc` — Label-Based Indexing

````python
df_idx = pd.DataFrame({
    "Name"  : ["Alice", "Bob", "Charlie", "Diana", "Eve"],
    "Age"   : [25, 30, 22, 28, 35],
    "Score" : [88.5, 92.0, 75.0, 95.5, 80.0],
    "City"  : ["NY", "LA", "NY", "Chicago", "NY"],
    "Dept"  : ["DS", "ML", "DS", "AI", "ML"]
}, index=["r0", "r1", "r2", "r3", "r4"])

print("DataFrame with custom index:\n", df_idx)
````

````python
# ── loc: Access single row ────────────────────────────────────────
print("\n--- loc: Access single row 'r1' ---")
print(df_idx.loc["r1"])

# ── loc: Access single column ────────────────────────────────────
print("\n--- loc: Access single column 'Score' ---")
print(df_idx.loc[:, "Score"])
````

````python
# ── loc: Access multiple rows & columns ──────────────────────────
print("\n--- loc: Multiple rows ['r0','r2'] & columns ['Name','Score'] ---")
print(df_idx.loc[["r0", "r2"], ["Name", "Score"]])
````

````python
# ── loc: Slice of rows ────────────────────────────────────────────

print("\n--- loc: Row slice r1 to r3 ---")
print(df_idx.loc["r1":"r3"])
````

````python
# ── loc: Conditional (boolean) selection ─────────────────────────
print("\n--- loc: Score > 85 ---")
print(df_idx.loc[df_idx["Score"] > 85])

print("\n--- loc: Score > 80 AND City == 'NY' ---")
print(df_idx.loc[(df_idx["Score"] > 80) & (df_idx["City"] == "NY")])
````

### `iloc` — Position-Based Indexing

````python
# ── iloc: Access single row by position ──────────────────────────
print("--- iloc: Single row at position 0 ---")
print(df_idx.iloc[0])

print("\n--- iloc: Single row at position -1 (last row) ---")
print(df_idx.iloc[-1])

# ── iloc: Access single column by position ────────────────────────
print("\n--- iloc: Column at position 2 (Score) ---")
print(df_idx.iloc[:, 2])
````

````python
# ── iloc: Subset of rows and columns ─────────────────────────────
print("\n--- iloc: Rows 0-2, Columns 0-2 ---")
print(df_idx.iloc[0:3, 0:3])

print("\n--- iloc: Every other row, first 3 columns ---")
print(df_idx.iloc[::2, :3])

print("\n--- iloc: Specific rows [0,2,4] and columns [0,2] ---")
print(df_idx.iloc[[0, 2, 4], [0, 2]])
````

## 12. DataFrame Slicing

### Types of Slicing
| Type | Description |
|------|-------------|
| Row slicing | Select a range of rows |
| Column slicing | Select specific columns by label or position |
| Combined slicing | Select both rows and columns simultaneously |

````python
df_sl = pd.DataFrame({
    "Name"  : ["Alice", "Bob", "Charlie", "Diana", "Eve"],
    "Age"   : [25, 30, 22, 28, 35],
    "Score" : [88.5, 92.0, 75.0, 95.5, 80.0],
    "City"  : ["NY", "LA", "NY", "Chicago", "NY"],
    "Dept"  : ["DS", "ML", "DS", "AI", "ML"]
})
print("Original DataFrame:\n", df_sl)
````

````python
# ── Row slicing ───────────────────────────────────────────────────
print("\n--- Row slicing: rows 1 to 3 (iloc) ---")
print(df_sl.iloc[1:4])
````

````python
# ── Column slicing — label-based ──────────────────────────────────
print("\n--- Column slicing (label-based): Name and Score ---")
print(df_sl[["Name", "Score"]])

# ── Column slicing — position-based ──────────────────────────────
print("\n--- Column slicing (position-based): columns 0,1,2 ---")
print(df_sl.iloc[:, 0:3])
````

````python
# ── Combined slicing — loc (label-based) ─────────────────────────
print("\n--- Combined slicing (loc — label-based rows+cols): rows 1-3, Name/Score/City ---")
print(df_sl.loc[1:3, ["Name", "Score", "City"]])

# ── Combined slicing — iloc (position-based) ─────────────────────
print("\n--- Combined slicing (iloc — position-based): rows 0-2, cols 0,2,3 ---")
print(df_sl.iloc[0:3, [0, 2, 3]])
````

## 13. `set_index` and Hierarchical Indexing

### Characteristics of `set_index`
| Feature | Description |
|---------|-------------|
| **Custom indexing** | Replace default RangeIndex with a column |
| **Hierarchical indexing** | Pass a list of columns to create a MultiIndex |
| **In-place operation** | Use `inplace=True` to modify without reassignment |

````python
df_si = pd.DataFrame({
    "Name"  : ["Alice", "Bob", "Charlie", "Diana", "Eve"],
    "Dept"  : ["DS", "ML", "DS", "AI", "ML"],
    "City"  : ["NY", "LA", "NY", "Chicago", "NY"],
    "Score" : [88.5, 92.0, 75.0, 95.5, 80.0]
})
print("Original:\n", df_si)
````

````python
# ── Custom indexing (single column) ──────────────────────────────
df_name_idx = df_si.set_index("Name")

print("\nCustom indexing — set_index('Name'):")
print(df_name_idx)
print("Access Alice:", df_name_idx.loc["Alice"])
````

````python
# ── Hierarchical (MultiIndex) ─────────────────────────────────────

df_multi = df_si.set_index(["Dept", "Name"])
print("\nHierarchical indexing — set_index(['Dept','Name']):")
print(df_multi)
print("\nAccess Dept='DS':", df_multi.loc["DS"])
print("\nAccess DS/Charlie:", df_multi.loc[("DS", "Charlie")])
````

````python
# ── in-place operation ────────────────────────────────────────────
df_inplace = df_si.copy()
df_inplace.set_index("Name", inplace=True)
print("\ninplace=True — original variable is modified:")
print(df_inplace.head())
````

## 14. Data Cleaning

### Topics Covered
1. Handling missing data — `isnull()`, `dropna()`, `fillna()`
2. Data type conversion — `astype()`
3. Rename rows
4. Handle duplicates — `drop_duplicates()`
5. String operations — `str` accessor

````python
df_clean = pd.DataFrame({
    "Name"  : ["Alice", "Bob", None, "Diana", "Eve", "Bob"],
    "Age"   : [25, np.nan, 22, 28, 35, 30],
    "Score" : [88.5, 92.0, np.nan, 95.5, np.nan, 92.0],
    "City"  : ["NY", "LA", "NY", None, "NY", "LA"],
    "Grade" : ["  A  ", "  A+ ", "  B+ ", "  A+ ", "  B  ", "  A+ "]
})

print("DataFrame with missing values:\n", df_clean)
````

````python
# ── isnull() / notnull() ──────────────────────────────────────────

print("\nisnull():\n", df_clean.isnull())
print("\nMissing value count per column:\n", df_clean.isnull().sum())
````

````python
# ── dropna() ─────────────────────────────────────────────────────
print("\ndropna() — drop rows with ANY missing value:")
print(df_clean.dropna())

print("\ndropna(axis=1) — drop COLUMNS with any missing value:")
print(df_clean.dropna(axis=1))

print("\ndropna(thresh=4) — keep rows with at least 4 non-null values:")
print(df_clean.dropna(thresh=4))
````

````python
# ── fillna() ─────────────────────────────────────────────────────
print("\nfillna(0) — fill all NaN with 0:")
print(df_clean.fillna(0))

print("\nfillna(method='ffill') — forward fill:")
print(df_clean.fillna(method='ffill'))

print("\nfillna(method='bfill') — backward fill:")
print(df_clean.fillna(method='bfill'))

print("\nfillna per-column dict:")
print(df_clean.fillna({"Age": df_clean["Age"].mean(), "Score": 0, "City": "Unknown"}))
````

````python
df_c2 = pd.DataFrame({
    "Name"  : ["Alice", "Bob", "Bob", "Diana", "Alice"],
    "Age"   : ["25", "30", "30", "28", "25"],
    "Score" : [88.5, 92.0, 92.0, 95.5, 88.5],
    "Grade" : ["  A  ", "  A+  ", "  A+  ", "  A+  ", "  A  "]
})

# ── astype() — data type conversion ──────────────────────────────
print("dtypes before:\n", df_c2.dtypes)
df_c2["Age"] = df_c2["Age"].astype(int)

print("\nAfter Age astype(int):", df_c2["Age"].dtype)
df_c2["Score"] = df_c2["Score"].astype(float)
df_c2["Name"]  = df_c2["Name"].astype("category")
print("After Name astype('category'):", df_c2["Name"].dtype)
````

````python
# ── Rename rows ───────────────────────────────────────────────────

df_row = df_c2.copy()
df_row = df_row.rename(index={0: "r0", 1: "r1", 2: "r2", 3: "r3", 4: "r4"})
print("\nRename rows:\n", df_row.index.tolist())
````

````python
# ── drop_duplicates() ────────────────────────────────────────────

print("\nOriginal (with duplicates):\n", df_c2[["Name","Age","Score"]])
print("\ndrop_duplicates() — keep first:")
print(df_c2.drop_duplicates())
print("\ndrop_duplicates(subset=['Name']) — based on Name only:")
print(df_c2.drop_duplicates(subset=["Name"]))
print("\ndrop_duplicates(keep='last'):")
print(df_c2.drop_duplicates(keep="last"))
````

## 16. Iteration Over Rows and Columns

> **Instructor Note:** Explicit loops over DataFrame rows are **slow**. Always prefer vectorized pandas operations. Use `iterrows()` / `itertuples()` only when necessary (e.g., row-by-row conditional logic that can't be vectorized).

| Method | Returns | Speed |
|--------|---------|-------|
| `iterrows()` | `(index, Series)` pairs | Slow |
| `itertuples()` | Named tuples | Faster than iterrows |
| Column loop | Column Series | Medium |
| **Vectorized** | Whole Series/array | **Fastest** |

````python
df_iter = pd.DataFrame({
    "Name" : ["Alice", "Bob", "Charlie"],
    "Age"  : [25, 30, 22],
    "Score": [88.5, 92.0, 75.0]
})

df_iter
````

````python
# ── iterrows() ───────────────────────────────────────────────────

print("--- iterrows() — (index, Series) pairs ---")
for idx, row in df_iter.iterrows():
    print(f"  Row {idx}: {row['Name']}, Age={row['Age']}, Score={row['Score']}")
````

````python
# ── itertuples() ─────────────────────────────────────────────────

print("\n--- itertuples() — named tuples (faster) ---")
for row in df_iter.itertuples():
    print(f"  Index={row.Index}, Name={row.Name}, Score={row.Score}")
````

````python
# ── Iterate over columns ─────────────────────────────────────────

print("\n--- Iterate over columns ---")
for col_name in df_iter.columns:
    print(f"  Column '{col_name}': {df_iter[col_name].tolist()}")
````

````python

# ── Preferred: vectorized alternative ────────────────────────────
print("\n--- Vectorized (preferred) — add bonus without loop ---")
df_iter["Bonus"]      = df_iter["Score"] * 0.1
df_iter["FinalScore"] = df_iter["Score"] + df_iter["Bonus"]
print(df_iter)
````

## 19. Merging DataFrames — `pd.merge()`

### Types of Join
| Join | Description |
|------|-------------|
| **inner** | Only matching keys in both tables |
| **left**  | All rows from left, matching from right |
| **right** | All rows from right, matching from left |
| **outer** | All rows from both, NaN where no match |

````python
employees = pd.DataFrame({
    "EmpID" : [1, 2, 3, 4, 5],
    "Name"  : ["Alice", "Bob", "Charlie", "Diana", "Eve"],
    "DeptID": [10, 20, 10, 30, 20]
})

departments = pd.DataFrame({
    "DeptID"  : [10, 20, 40],
    "DeptName": ["Data Science", "Machine Learning", "DevOps"]
})

print("Employees:\n", employees)
print("\nDepartments:\n", departments)
````

````python
# ── inner merge ───────────────────────────────────────────────────
print("\n--- INNER merge (only matching DeptIDs) ---")
print(pd.merge(employees, departments, on="DeptID", how="inner"))
````

````python
# ── left merge ────────────────────────────────────────────────────
print("\n--- LEFT merge (all employees kept) ---")
print(pd.merge(employees, departments, on="DeptID", how="left"))
````

````python
# ── right merge ───────────────────────────────────────────────────
print("\n--- RIGHT merge (all departments kept) ---")
print(pd.merge(employees, departments, on="DeptID", how="right"))
````

````python
# ── outer merge ───────────────────────────────────────────────────
print("\n--- OUTER merge (everything from both) ---")
print(pd.merge(employees, departments, on="DeptID", how="outer"))
````

## 21. GroupBy Operations

### The Split → Apply → Combine Pattern

```
Original DataFrame
       ↓
   SPLIT (group by key)
   ┌──────┬──────┐
   │ DS   │ ML   │  ← groups
   └──────┴──────┘
       ↓
   APPLY (aggregation / transform / filter)
       ↓
   COMBINE (assemble result)
```

| Step | Description |
|------|-------------|
| **Split** | Divide DataFrame into groups by one or more keys |
| **Apply** | Apply a function to each group independently |
| **Combine** | Merge the results into a single DataFrame/Series |

````python
df_gb = pd.DataFrame({
    "Name"  : ["Alice","Bob","Charlie","Diana","Eve","Frank","Grace","Henry"],
    "Dept"  : ["DS","ML","DS","AI","ML","DS","AI","ML"],
    "City"  : ["NY","LA","NY","Chicago","NY","LA","Chicago","NY"],
    "Score" : [88, 92, 75, 95, 80, 70, 85, 90],
    "Salary": [70000, 80000, 65000, 90000, 75000, 60000, 85000, 78000],
    "Exp"   : [3, 5, 2, 7, 4, 1, 6, 4]
})

print("Base DataFrame:\n", df_gb)
````

````python
# ── Basic GroupBy ─────────────────────────────────────────────────

print("\n--- GroupBy single column (Dept) — mean Score ---")
print(df_gb.groupby("Dept")["Score"].mean())
````

````python
# ── GroupBy on multiple columns ────────────────────────────────────

print("\n--- GroupBy multiple columns (Dept + City) — mean Score ---")
print(df_gb.groupby(["Dept","City"])["Score"].mean())
````

````python
# ── Multiple aggregation functions ────────────────────────────────

print("\n--- agg() — multiple functions on Score ---")
print(df_gb.groupby("Dept")["Score"].agg(["mean","min","max","std","count"]))

print("\n--- agg() — different functions per column ---")
print(df_gb.groupby("Dept").agg({
    "Score" : ["mean","max"],
    "Salary": ["mean","sum"],
    "Exp"   : "mean"
}))
````

````python
# ── filter() — keep groups satisfying a condition ─────────────────
print("--- filter() — keep Depts where mean Score > 82 ---")

filtered = df_gb.groupby("Dept").filter(lambda g: g["Score"].mean() > 82)
print(filtered)
````

````python
# ── transform() — group-wise transformation (preserves shape) ─────

print("\n--- transform() — normalize Score within each Dept (z-score) ---")
df_gb["Score_zscore"] = df_gb.groupby("Dept")["Score"].transform(
    lambda x: (x - x.mean()) / x.std()
)
print(df_gb[["Name","Dept","Score","Score_zscore"]].round(3))

print("\n--- transform('mean') — fill each row with its group mean ---")
df_gb["Dept_AvgScore"] = df_gb.groupby("Dept")["Score"].transform("mean")
print(df_gb[["Name","Dept","Score","Dept_AvgScore"]])
````

````python
# ── GroupBy with custom function ──────────────────────────────────
print("\n--- Custom agg function — Score range (max-min) per Dept ---")
def score_range(series):
    return series.max() - series.min()

print(df_gb.groupby("Dept")["Score"].agg(score_range))
````

````python
# ── Named aggregation (pandas >= 0.25) ────────────────────────────
print("\n--- Named aggregation ---")
result = df_gb.groupby("Dept").agg(
    avg_score   = ("Score",  "mean"),
    total_salary= ("Salary", "sum"),
    headcount   = ("Name",   "count")
)
print(result)
````

## 22. Pivot Tables — `pd.pivot_table()`

### What is a Pivot Table?
A **Pivot Table** reshapes data by aggregating values across two axes.  
Think of it as a powerful GroupBy with a 2D grid output.

```
pd.pivot_table(df,
    values  = 'column_to_aggregate',
    index   = 'rows',
    columns = 'columns',
    aggfunc = 'mean'   # default
)
```

````python
df_pv = pd.DataFrame({
    "Name"  : ["Alice","Bob","Charlie","Diana","Eve","Frank","Grace","Henry","Iris","Jack"],
    "Dept"  : ["DS","ML","DS","AI","ML","DS","AI","ML","DS","AI"],
    "City"  : ["NY","LA","NY","Chicago","NY","LA","Chicago","NY","LA","NY"],
    "Score" : [88, 92, 75, 95, 80, 70, 85, 90, 78, 88],
    "Salary": [70000, 80000, 65000, 90000, 75000, 60000, 85000, 78000, 72000, 82000],
    "Quarter": ["Q1","Q2","Q1","Q2","Q1","Q2","Q1","Q2","Q1","Q2"]
})

df_pv
````

````python
# ── Basic pivot table ─────────────────────────────────────────────
print("--- Basic pivot table: mean Score by Dept × City ---")
pt1 = pd.pivot_table(df_pv, values="Score", index="Dept", columns="City", aggfunc="mean")
print(pt1)
````

````python
# ── Multiple aggregation functions ────────────────────────────────
print("\n--- Multiple aggfunc: mean and max Score ---")
pt2 = pd.pivot_table(df_pv, values="Score", index="Dept", columns="City",
                     aggfunc=["mean","max"])
print(pt2)
````

````python
# ── Handling missing values with fill_value ────────────────────────
print("\n--- Handle NaN with fill_value=0 ---")
pt3 = pd.pivot_table(df_pv, values="Score", index="Dept", columns="City",
                     aggfunc="mean", fill_value=0)
print(pt3)
````

````python
# ── Adding margins (row/column subtotals) ─────────────────────────
print("\n--- margins=True adds row/column totals ---")
pt4 = pd.pivot_table(df_pv, values="Score", index="Dept", columns="City",
                     aggfunc="mean", fill_value=0, margins=True, margins_name="TOTAL")
print(pt4)
````

````python
# ── Multi-level pivot table ───────────────────────────────────────
print("\n--- Multi-level pivot: index=[Dept,Quarter], columns=City ---")
pt5 = pd.pivot_table(df_pv, values="Score", index=["Dept","Quarter"],
                     columns="City", aggfunc="mean", fill_value=0)
print(pt5)
````

## 23. Working with DateTime

### DateTime Data Types in Pandas

| Type | Description | Example |
|------|-------------|---------|
| `datetime64[ns]` | Absolute timestamp | `2024-01-15 08:30:00` |
| `timedelta64[ns]` | Duration / difference between times | `3 days 04:00:00` |

### The `.dt` Accessor
Just like `.str` for strings, `.dt` gives access to datetime properties on a Series.

````python
# ── Converting strings to datetime ───────────────────────────────
dates_str = pd.Series(["2024-01-15", "2024-03-22", "2024-07-04",
                        "2024-11-28", "2025-02-14"])
dates = pd.to_datetime(dates_str)
print("Converted to datetime:\n", dates)
print("dtype:", dates.dtype)
````

````python
# ── Custom format string ──────────────────────────────────────────
custom_dates = pd.to_datetime(["15/01/2024", "22/03/2024"], format="%d/%m/%Y")
print("\nCustom format parsing:", custom_dates)
````

````python
# ── Extracting date components with .dt accessor ──────────────────
df_dt = pd.DataFrame({
    "Name"      : ["Alice","Bob","Charlie","Diana","Eve"],
    "JoinDate"  : pd.to_datetime(["2022-03-15","2020-07-01","2023-01-20","2019-11-30","2024-06-05"]),
    "BirthDate" : pd.to_datetime(["1995-08-22","1988-04-10","1997-12-05","1992-06-15","2000-09-30"])
})

print("\nDate components extracted with .dt:\n")
df_dt["Year"]    = df_dt["JoinDate"].dt.year
df_dt["Month"]   = df_dt["JoinDate"].dt.month
df_dt["Day"]     = df_dt["JoinDate"].dt.day
df_dt["Weekday"] = df_dt["JoinDate"].dt.weekday   # 0=Monday
df_dt["DayName"] = df_dt["JoinDate"].dt.day_name()
df_dt["Quarter"] = df_dt["JoinDate"].dt.quarter
print(df_dt[["Name","JoinDate","Year","Month","Day","Weekday","DayName","Quarter"]])
````

````python
# ── Creating date ranges ─────────────────────────────────────────
print("--- pd.date_range: daily frequency ---")
dr_daily = pd.date_range(start="2024-01-01", end="2024-01-10", freq="D")
print(dr_daily)

print("\n--- pd.date_range: monthly frequency (6 periods) ---")
dr_monthly = pd.date_range(start="2024-01-01", periods=6, freq="ME")
print(dr_monthly)

print("\n--- pd.date_range: business days ---")
dr_biz = pd.date_range(start="2024-01-01", periods=5, freq="B")
print(dr_biz)

print("\n--- pd.date_range: hourly ---")
dr_hourly = pd.date_range(start="2024-01-01 08:00", periods=5, freq="h")
print(dr_hourly)
````

````python
# ── Timedelta operations ──────────────────────────────────────────
print("\n--- Timedelta: calculate tenure in days ---")
df_dt["Today"]  = pd.Timestamp("2024-12-31")
df_dt["Tenure"] = df_dt["Today"] - df_dt["JoinDate"]
print(df_dt[["Name","JoinDate","Tenure"]])

print("\n--- Extract days from timedelta using .dt.days ---")
df_dt["TenureDays"]  = df_dt["Tenure"].dt.days
df_dt["TenureYears"] = (df_dt["TenureDays"] / 365).round(1)
print(df_dt[["Name","TenureDays","TenureYears"]])

print("\n--- Adding timedelta offsets ---")
df_dt["ContractEnd"] = df_dt["JoinDate"] + pd.Timedelta(days=365)
print(df_dt[["Name","JoinDate","ContractEnd"]])
````

## 25. Handling Large Datasets

### Why It Matters
In real-world ML projects, datasets can be **gigabytes** in size. Loading them naively into memory crashes your kernel.

### Strategies
| Strategy | Tool | Description |
|----------|------|-------------|
| **Chunking** | `chunksize` | Read file in pieces |
| **Optimize dtypes** | `astype()` | Downcast int/float, use category |
| **Monitor memory** | `memory_usage()` | Track DataFrame RAM usage |
| **Partial load** | `nrows` | Load only first N rows for exploration |

````python
# ── Generate a large sample CSV for demonstration ─────────────────
large_csv = r"c:\learning\AIML_learning\pandas\large_sample.csv"

np.random.seed(42)
n = 100_000
depts  = np.random.choice(["DS","ML","AI","DevOps","Cloud"], n)
cities = np.random.choice(["NY","LA","Chicago","Houston","Phoenix"], n)

df_large_gen = pd.DataFrame({
    "EmpID"  : np.arange(1, n+1),
    "Name"   : [f"Emp_{i}" for i in range(n)],
    "Dept"   : depts,
    "City"   : cities,
    "Age"    : np.random.randint(22, 60, n),
    "Score"  : np.random.uniform(50, 100, n).round(2),
    "Salary" : np.random.randint(40000, 120000, n),
    "Exp"    : np.random.randint(0, 20, n)
})
df_large_gen.to_csv(large_csv, index=False)
print(f"Generated {n:,} row CSV at: {large_csv}")
print(f"   File size: {os.path.getsize(large_csv)/1024:.1f} KB")
````

````python
# ── nrows: load first N rows for quick inspection ─────────────────
print("--- nrows=5: quick preview ---")
df_preview = pd.read_csv(large_csv, nrows=5)
print(df_preview)

# ── Chunking: process file in pieces ──────────────────────────────
print("\n--- Chunking: compute dept-wise mean salary without loading full file ---")
chunk_size = 10_000
dept_salary_totals = {}
dept_counts        = {}

for chunk in pd.read_csv(large_csv, chunksize=chunk_size):
    for dept, grp in chunk.groupby("Dept"):
        dept_salary_totals[dept] = dept_salary_totals.get(dept, 0) + grp["Salary"].sum()
        dept_counts[dept]        = dept_counts.get(dept, 0) + len(grp)

dept_avg = {d: dept_salary_totals[d] / dept_counts[d] for d in dept_salary_totals}
print("Dept-wise mean Salary (via chunking):")
for d, avg in sorted(dept_avg.items()):
    print(f"  {d}: ${avg:,.0f}")
````

````python
# ── Memory monitoring ─────────────────────────────────────────────
df_mem = pd.read_csv(large_csv)
print("--- memory_usage(deep=True) BEFORE optimization ---")
mem_before = df_mem.memory_usage(deep=True)
print(mem_before)
total_before = mem_before.sum() / 1024**2
print(f"\nTotal memory BEFORE: {total_before:.2f} MB")

# ── Dtype optimization ────────────────────────────────────────────
print("\n--- Optimizing dtypes ---")
print("BEFORE dtypes:\n", df_mem.dtypes)

df_opt = df_mem.copy()

# object → category (low-cardinality columns)
for col in ["Dept", "City"]:
    df_opt[col] = df_opt[col].astype("category")

# int64 → int32 (safe if values fit in int32 range)
for col in ["EmpID", "Age", "Salary", "Exp"]:
    df_opt[col] = df_opt[col].astype("int32")

# float64 → float32
df_opt["Score"] = df_opt["Score"].astype("float32")

print("\nAFTER dtypes:\n", df_opt.dtypes)

mem_after = df_opt.memory_usage(deep=True)
total_after = mem_after.sum() / 1024**2
print(f"\nTotal memory BEFORE : {total_before:.2f} MB")
print(f"Total memory AFTER  : {total_after:.2f} MB")
print(f"Memory SAVED        : {total_before - total_after:.2f} MB ({(1 - total_after/total_before)*100:.1f}% reduction)")
````
